### Exercício 2: Eliminação de Gauss Manual

Sistema a resolver:
$$
\begin{cases}
2x + y - z = 8 \\
-3x - y + 2z = -11 \\
-2x + y + 2z = -3
\end{cases}
$$

#### Passo 1: Montagem da Matriz Aumentada $[A|\mathbf{b}]$
$$
[A|\mathbf{b}] = \left[\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
-3 & -1 & 2 & -11 \\
-2 & 1 & 2 & -3
\end{array}\right]
$$

#### Passo 2: Eliminação na Coluna 1 (Pivô $a_{11} = 2$)
- Linha 2: Multiplicador $m_{21} = \frac{-3}{2} = -1.5$. Operação: $L_2 \leftarrow L_2 - (-1.5)L_1 = L_2 + 1.5 L_1$
  - Elemento $(2,1): -3 + 1.5(2) = 0$
  - Elemento $(2,2): -1 + 1.5(1) = 0.5$
  - Elemento $(2,3): 2 + 1.5(-1) = 0.5$
  - Termo $(2,4): -11 + 1.5(8) = 1$
- Linha 3: Multiplicador $m_{31} = \frac{-2}{2} = -1$. Operação: $L_3 \leftarrow L_3 - (-1)L_1 = L_3 + L_1$
  - Elemento $(3,1): -2 + 2 = 0$
  - Elemento $(3,2): 1 + 1 = 2$
  - Elemento $(3,3): 2 + (-1) = 1$
  - Termo $(3,4): -3 + 8 = 5$

Matriz após eliminação na coluna 1:
$$
\left[\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
0 & 0.5 & 0.5 & 1 \\
0 & 2 & 1 & 5
\end{array}\right]
$$

#### Passo 3: Eliminação na Coluna 2 (Pivô $a_{22} = 0.5$)
- Linha 3: Multiplicador $m_{32} = \frac{2}{0.5} = 4$. Operação: $L_3 \leftarrow L_3 - 4 L_2$
  - Elemento $(3,1): 0 - 4(0) = 0$
  - Elemento $(3,2): 2 - 4(0.5) = 0$
  - Elemento $(3,3): 1 - 4(0.5) = -1$
  - Termo $(3,4): 5 - 4(1) = 1$

Matriz escalonada final:
$$
\left[\begin{array}{ccc|c}
2 & 1 & -1 & 8 \\
0 & 0.5 & 0.5 & 1 \\
0 & 0 & -1 & 1
\end{array}\right]
$$

#### Passo 4: Substituição Regressiva
1. Da 3ª linha:
   $$ -z = 1 \implies z = -1 $$
2. Da 2ª linha:
   $$ 0.5y + 0.5z = 1 \implies 0.5y + 0.5(-1) = 1 \implies 0.5y = 1.5 \implies y = 3 $$
3. Da 1ª linha:
   $$ 2x + y - z = 8 \implies 2x + 3 - (-1) = 8 \implies 2x + 4 = 8 \implies 2x = 4 \implies x = 2 $$

**Solução:** $(x, y, z) = (2, 3, -1)$

### Exercício 3: Determinação do Número de Soluções

A classificação é obtida escalonando a matriz aumentada $[A|\mathbf{b}]$ e comparando o posto da matriz dos coeficientes $\text{posto}(A)$ com o posto da matriz aumentada $\text{posto}(A|\mathbf{b})$.



#### a) Sistema:
$$
\begin{cases}
x + y = 3 \\
2x + 2y = 6
\end{cases}
$$

Matriz aumentada:
$$
[A|\mathbf{b}] = \left[\begin{array}{cc|c}
1 & 1 & 3 \\
2 & 2 & 6
\end{array}\right]
$$
Operação: $L_2 \leftarrow L_2 - 2L_1$:
$$
\left[\begin{array}{cc|c}
1 & 1 & 3 \\
0 & 0 & 0
\end{array}\right]
$$

- $\text{posto}(A) = 1$ e $\text{posto}(A|\mathbf{b}) = 1$.
- Como o posto é menor que o número de incógnitas ($n = 2$), o sistema é **Possível e Indeterminado (SPI)**.
- **Conclusão:** O sistema possui **infinitas soluções** (geometria: retas coincidentes), com solução geral dada por $(x, 3 - x), \forall x \in \mathbb{R}$.

---

#### b) Sistema:
$$
\begin{cases}
x + y = 3 \\
2x + 2y = 7
\end{cases}
$$

Matriz aumentada:
$$
[A|\mathbf{b}] = \left[\begin{array}{cc|c}
1 & 1 & 3 \\
2 & 2 & 7
\end{array}\right]
$$
Operação: $L_2 \leftarrow L_2 - 2L_1$:
$$
\left[\begin{array}{cc|c}
1 & 1 & 3 \\
0 & 0 & 1
\end{array}\right]
$$

- $\text{posto}(A) = 1$ e $\text{posto}(A|\mathbf{b}) = 2$.
- A última linha representa a equação $0x + 0y = 1 \implies 0 = 1$ (contradição matemática).
- **Conclusão:** O sistema é **Impossível (SI)**, ou seja, possui **nenhuma solução** (geometria: retas paralelas e distintas).

---

#### c) Sistema:
$$
\begin{cases}
x + y = 3 \\
2x - y = 0
\end{cases}
$$

Matriz aumentada:
$$
[A|\mathbf{b}] = \left[\begin{array}{cc|c}
1 & 1 & 3 \\
2 & -1 & 0
\end{array}\right]
$$
Operação: $L_2 \leftarrow L_2 - 2L_1$:
$$
\left[\begin{array}{cc|c}
1 & 1 & 3 \\
0 & -3 & -6
\end{array}\right]
$$

- $\text{posto}(A) = 2$ e $\text{posto}(A|\mathbf{b}) = 2$.
- Como $\text{posto}(A) = \text{posto}(A|\mathbf{b}) = n = 2$, o sistema é **Possível e Determinado (SPD)**.
- Substituição: $-3y = -6 \implies y = 2$; $x + 2 = 3 \implies x = 1$.
- **Conclusão:** O sistema possui **solução única**: $(x, y) = (1, 2)$ (geometria: retas concorrentes no ponto $(1, 2)$).



```python
# Verificação rápida em Python dos três sistemas
import numpy as np

sistemas = {
    "a) Infinitas Soluções (SPI)": (np.array([[1., 1.], [2., 2.]]), np.array([3., 6.])),
    "b) Nenhuma Solução (SI)":    (np.array([[1., 1.], [2., 2.]]), np.array([3., 7.])),
    "c) Solução Única (SPD)":      (np.array([[1., 1.], [2., -1.]]), np.array([3., 0.]))
}

for nome, (A, b) in sistemas.items():
    Ab = np.column_stack([A, b])
    posto_A = np.linalg.matrix_rank(A)
    posto_Ab = np.linalg.matrix_rank(Ab)
    n = A.shape[1]
    
    if posto_A < posto_Ab:
        status = "Nenhuma solução (Sistema Impossível - SI)"
    elif posto_A == posto_Ab < n:
        status = "Infinitas soluções (Sistema Possível e Indeterminado - SPI)"
    else:
        status = "Solução única (Sistema Possível e Determinado - SPD)"
        
    print(f"{nome}:")
    print(f"  Posto(A) = {posto_A} | Posto(A|b) = {posto_Ab} | Incógnitas = {n}")
    print(f"  Classificação: {status}
")

```

## Parte 3 — Implementação Numérica em Python

### Exercício 4: Algoritmo de Eliminação de Gauss sem Pivoteamento



```python
def eliminacao_gauss(A, b, verbose=True):
    '''
    Resolve o sistema linear Ax = b utilizando Eliminação de Gauss sem pivoteamento.
    
    Parâmetros:
        A (array-like): Matriz dos coeficientes (n x n).
        b (array-like): Vetor dos termos independentes (n).
        verbose (bool): Se True, exibe as etapas intermediárias da matriz aumentada.
        
    Retorna:
        x (ndarray): Vetor solução.
    '''
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float).reshape(-1, 1)
    n = len(b)
    
    # 1. Formar a matriz aumentada [A|b]
    Ab = np.hstack([A, b])
    if verbose:
        print("=== Matriz Aumentada Inicial [A|b] ===")
        print(np.round(Ab, 4))
        print()

    # 2. Eliminação dos elementos abaixo dos pivôs
    for i in range(n - 1):
        pivo = Ab[i, i]
        if abs(pivo) < 1e-12:
            raise ValueError(f"Pivô nulo ou muito próximo de zero na linha {i+1} (a_{i+1}{i+1} = {pivo}).")
            
        for j in range(i + 1, n):
            m = Ab[j, i] / pivo
            Ab[j, i:] -= m * Ab[i, i:]
            
        if verbose:
            print(f"--- Após eliminação na coluna {i+1} (pivô = {pivo:.4f}) ---")
            print(np.round(Ab, 4))
            print()

    # 3. Substituição regressiva
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        soma = np.sum(Ab[i, i + 1:n] * x[i + 1:n])
        x[i] = (Ab[i, -1] - soma) / Ab[i, i]

    if verbose:
        print("=== Solução Calculada ===")
        for idx, val in enumerate(x):
            print(f"x_{idx+1} = {val:.6f}")
        print()

    return x

```

```python
# Teste da função com o sistema do Exercício 2
A2 = np.array([
    [ 2.0,  1.0, -1.0],
    [-3.0, -1.0,  2.0],
    [-2.0,  1.0,  2.0]
])

b2 = np.array([8.0, -11.0, -3.0])

# Resolução pelo algoritmo implementado
solucao_gauss = eliminacao_gauss(A2, b2, verbose=True)

# Resolução pelo NumPy para comparação
solucao_numpy = np.linalg.solve(A2, b2)

# Verificação do resíduo ||Ax - b||
residuo = np.linalg.norm(np.dot(A2, solucao_gauss) - b2)

print("Comparação dos resultados:")
print(f"  Implementação manual: {solucao_gauss}")
print(f"  numpy.linalg.solve:   {solucao_numpy}")
print(f"  Diferença máxima:     {np.max(np.abs(solucao_gauss - solucao_numpy)):.2e}")
print(f"  Norma do resíduo ||Ax - b||: {residuo:.2e}")

```

## Parte 4 — Aplicação: Modelo de Balanço Térmico

### Exercício 5: Rede Térmica Simplificada

Sistema térmico que rege as temperaturas desconhecidas $T_1, T_2, T_3$:
$$
\begin{bmatrix}
4 & -1 & 0 \\
-1 & 4 & -1 \\
0 & -1 & 3
\end{bmatrix}
\begin{bmatrix}
T_1 \\
T_2 \\
T_3
\end{bmatrix}
=
\begin{bmatrix}
15 \\
10 \\
10
\end{bmatrix}
$$

---

### a) Resolução Manual

#### 1. Matriz Aumentada Inicial
$$
[A|\mathbf{b}] = \left[\begin{array}{ccc|c}
4 & -1 & 0 & 15 \\
-1 & 4 & -1 & 10 \\
0 & -1 & 3 & 10
\end{array}\right]
$$

#### 2. Eliminação na Coluna 1 (Pivô $a_{11} = 4$)
- Linha 2: $m_{21} = \frac{-1}{4} = -0.25$. Operação: $L_2 \leftarrow L_2 - (-0.25)L_1 = L_2 + 0.25 L_1$
  - $(2,1): -1 + 0.25(4) = 0$
  - $(2,2): 4 + 0.25(-1) = 3.75 = \frac{15}{4}$
  - $(2,3): -1 + 0.25(0) = -1 = -\frac{4}{4}$
  - $(2,4): 10 + 0.25(15) = 13.75 = \frac{55}{4}$
- Linha 3: Já possui zero na coluna 1.

Matriz resultante:
$$
\left[\begin{array}{ccc|c}
4 & -1 & 0 & 15 \\
0 & 3.75 & -1 & 13.75 \\
0 & -1 & 3 & 10
\end{array}\right]
$$

#### 3. Eliminação na Coluna 2 (Pivô $a_{22} = 3.75 = \frac{15}{4}$)
- Linha 3: $m_{32} = \frac{-1}{15/4} = -\frac{4}{15}$. Operação: $L_3 \leftarrow L_3 - \left(-\frac{4}{15}\right)L_2 = L_3 + \frac{4}{15}L_2$
  - $(3,1): 0$
  - $(3,2): -1 + \frac{4}{15}\left(\frac{15}{4}\right) = 0$
  - $(3,3): 3 + \frac{4}{15}(-1) = \frac{45}{15} - \frac{4}{15} = \frac{41}{15} \approx 2.7333$
  - $(3,4): 10 + \frac{4}{15}\left(\frac{55}{4}\right) = 10 + \frac{55}{15} = \frac{150 + 55}{15} = \frac{205}{15} = \frac{41}{3} \approx 13.6667$

Matriz escalonada final:
$$
\left[\begin{array}{ccc|c}
4 & -1 & 0 & 15 \\
0 & \frac{15}{4} & -1 & \frac{55}{4} \\
0 & 0 & \frac{41}{15} & \frac{41}{3}
\end{array}\right]
$$

#### 4. Substituição Regressiva
1. Da 3ª linha:
   $$ \frac{41}{15} T_3 = \frac{41}{3} \implies T_3 = \frac{41}{3} \cdot \frac{15}{41} = \frac{15}{3} \implies T_3 = 5 $$
2. Da 2ª linha:
   $$ \frac{15}{4} T_2 - T_3 = \frac{55}{4} \implies \frac{15}{4} T_2 - 5 = \frac{55}{4} \implies \frac{15}{4} T_2 = \frac{75}{4} \implies T_2 = \frac{75}{15} \implies T_2 = 5 $$
3. Da 1ª linha:
   $$ 4 T_1 - T_2 = 15 \implies 4 T_1 - 5 = 15 \implies 4 T_1 = 20 \implies T_1 = 5 $$

**Temperaturas:** $T_1 = 5^\circ\text{C}, \quad T_2 = 5^\circ\text{C}, \quad T_3 = 5^\circ\text{C}$



### b) Implementação em Python


```python
# Coeficientes da rede térmica
A_termico = np.array([
    [ 4.0, -1.0,  0.0],
    [-1.0,  4.0, -1.0],
    [ 0.0, -1.0,  3.0]
])

b_termico = np.array([15.0, 10.0, 10.0])

# Resolução pelo algoritmo implementado
T_gauss = eliminacao_gauss(A_termico, b_termico, verbose=True)

# Validação com NumPy
T_numpy = np.linalg.solve(A_termico, b_termico)

print("Resultado do Balanço Térmico:")
print(f"  T1 = {T_gauss[0]:.2f} °C")
print(f"  T2 = {T_gauss[1]:.2f} °C")
print(f"  T3 = {T_gauss[2]:.2f} °C")
print(f"  Validação com numpy.linalg.solve: {T_numpy}")

```

## Conclusão e Breve Interpretação dos Resultados

1. **Eliminação de Gauss:**
   - O algoritmo triangulariza a matriz através de combinações lineares sucessivas, reduzindo a complexidade de inversão matricial direta para uma substituição regressiva de complexidade $\mathcal{O}(n^2)$ após eliminação $\mathcal{O}(n^3)$.
   - Os resultados da implementação manual coincidem exatamente com a rotina `numpy.linalg.solve` (diferença da ordem de $10^{-16}$).

2. **Propriedades dos Sistemas:**
   - O número de soluções é ditado pelos postos de $A$ e $[A|\mathbf{b}]$ (Teorema de Rouché-Capelli):
     - $\text{posto}(A) = \text{posto}(A|\mathbf{b}) = n \implies$ **SPD** (solução única).
     - $\text{posto}(A) = \text{posto}(A|\mathbf{b}) < n \implies$ **SPI** (infinitas soluções).
     - $\text{posto}(A) < \text{posto}(A|\mathbf{b}) \implies$ **SI** (impossível).

3. **Balanço Térmico:**
   - A distribuição equilibrada de perdas e ganhos de calor pelos nós da rede resultou em uma temperatura uniforme de $5^\circ\text{C}$ para os três pontos ($T_1 = T_2 = T_3 = 5^\circ\text{C}$).
